# Penyelesaian Persamaan Non Linear Metode Bisection

```python
import numpy as np
```

```python
def baca_angka(pesan):
  while True:
    try:
      return float(input(pesan))
    except valueError:
      print("Input tidak valid, masukan angka.")
```

```python
def buat_fungsi(ekspresi):
    ekspresi = ekspresi.replace("^", "**")
    nama_aman = {
        "sin": np.sin, "cos": np.cos, "tan": np.tan,
        "exp": np.exp, "log": np.log, "sqrt": np.sqrt, "pi": np.pi,
    }

    def f(x):
        return eval(ekspresi, {"__builtins__": {}}, {**nama_aman, "x": x})

    return f
```

```python
def f(x):
  return eval(ekspresi, {"__Builtns__"})
  return eval(ekspresi, {"__builtins__": {}}, {**nama_aman, "x": x})

  return f
```

```python
def bisection(f, a, b, tol=1e-6, maks_iter=100):
    if f(a) * f(b) > 0:
        print("\nGagal: f(a) dan f(b) bertanda sama.")
        print("Tidak dijamin ada akar di interval ini, coba interval lain.")
        return None

    print(f"\n{'Iter':>4} {'a':>12} {'b':>12} {'c':>12} {'f(c)':>13} {'Galat':>11}")
    c = a
    for i in range(1, maks_iter + 1):
        c = (a + b) / 2
        fc = f(c)
        galat = abs(b - a) / 2
        print(f"{i:>4} {a:>12.6f} {b:>12.6f} {c:>12.6f} {fc:>13.6f} {galat:>11.2e}")

        if fc == 0 or galat < tol:
            return c

        if f(a) * fc < 0:
            b = c
        else:
            a = c

    print("\nPeringatan: batas iterasi tercapai sebelum toleransi terpenuhi.")
    return c
```

```python
def main():
    print("=== Metode Bisection ===")
    print("Contoh fungsi: x**3 - x - 2   atau   cos(x) - x\n")

    ekspresi = input("Masukkan f(x): ")
    try:
        f = buat_fungsi(ekspresi)
        f(1.0)
    except Exception as e:
        print(f"Ekspresi tidak valid: {e}")
        return

    a = baca_angka("Batas bawah interval (a): ")
    b = baca_angka("Batas atas interval (b): ")
    tol = baca_angka("Toleransi galat (contoh 0.000001): ")

    akar = bisection(f, a, b, tol)

    if akar is not None:
        print(f"\nAkar hampiran: x = {akar:.6f}")
        print(f"Nilai f(x)   : {f(akar):.2e}")


if __name__ == "__main__":
    main()
```
