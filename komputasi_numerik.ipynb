# **Program Python Untuk Persamaan Linear 3 Variabel**
**Dibawah ini merupakan source code mengimpor library Numpy.** numpy ini berfungsi untuk memberikan ruang bagi bahasa pemograman Python untuk dapat memproses array.

```python
import numpy as np
```

**Dibawah ini merupakan sebuah fungsi untuk membaca angka yang user masukan.**

```python
def baca_angka(pesan):
  while True:
    try:
      return float(input(pesan))
    except ValueError:
      print("Input tidak valid, masukan angka: ")
```

```python
def main():
  print("Penyelesaikan sistem persamaan linear 3 variabel")
  print("Bentuk persamaan: a*x + b*y + c*z = d")

  A = np.zeros((3, 3))
  b = np.zeros((3))

  for i in range(3):
    print(f"Persamaan {i + 1}:")
    for j, nama in enumerate ("xyz"):
      A[i, j] = baca_angka(f" koefisien {nama}: ")
    b[i] = baca_angka("Konstanta (ruas kanan): ")

  det = np.linalg.det(A)
  print(f"\nDeterminan matriks koefisien: {det:.4f}")

  if np.isclose(det, 0):
    print("Determinan = 0, sistem tidak memiliki solusi tunggal")
    print("(Bisa tidak ada solusi atau solusi tak hingga banyak)")
    return

  x, y, z = np.linalg.solve(A, b)

  print("\nSolusi:")
  print(f"x = {x:.4f}")
  print(f"y = {y:.4f}")
  print(f"z = {z:.4f}")

  cocok = np.allclose(A @ np.array([x, y, z]), b)
  print("\nVerifikasi (a.x = b):", "benar" if cocok else "tidak cocok")

if __name__ == "__main__":
  main()
```
